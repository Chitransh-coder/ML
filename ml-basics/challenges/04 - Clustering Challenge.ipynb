## Visualising dataset

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as mp

scaled = MinMaxScaler().fit_transform(data)
pca = PCA(n_components=2).fit(scaled)
vis = pca.transform(scaled)
mp.scatter(vis[:,0], vis[:,1])
mp.show()
```

## Doing K-Means Clustering

```python
from sklearn.cluster import KMeans

k = KMeans(n_clusters=3, init='k-means++', n_init=100, max_iter=1000).fit_predict(scaled)
cd = {0:'red', 1:'blue', 2:'green'}
md = {0:'*', 1:'x', 2:'+'}
mk = [md[i] for i in k]
mc = [cd[i] for i in k]
mp.scatter(vis[:,0], vis[:,1], marker='.', c=mc)
mp.show()
```

## Doing agglomerative clustering

```python
from sklearn.cluster import AgglomerativeClustering

A = AgglomerativeClustering(n_clusters=3).fit_predict(scaled)
mp.scatter(vis[:,0], vis[:,1], marker='.', c=[cd[i] for i in A])
mp.show()
```
