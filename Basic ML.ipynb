## Linear Algebra

Let's start with complete basics, with a **VECTOR**.

A **_vector_** is a representation of a point in space. A vector can be defined in 1D and 2D spaces also, but its real significance comes only when it is defined in a 3D space, i.e., using **x**, **y** & **z** coordinate system.

An important point to note is that these _x_, _y_ & _z_ coordinates may or may not be independant of each other. But, for learning's sake, let's assume that they are independant.
Since they carry no relation with each other, we can write them in a **_column matrix_**.

### What is a Matrix??

A **_matrix_** is a tabular representation of numbers, separated by space. These numbers, individually, may not contain any significance (not in our case, as they are coordinates) but inside a matrix, they carry many notable and important properties.

Let's take an example:

Assume your home is at the center as this is your current position. Let's set its coordinate to 0.
Now, you have to go to your school. Assuming your home and school lie on a straight line, the distance between your home and your school will be the co-ordinate of your school. Suppose the distance between your home and school is 4 km, its matrix representation will be as:

$$ [4] $$

But, in the real world, that is not the case, your home does not necessarily lie on the same straight line as your school, so what will be its coordinates??

Let's assume you have to travel 5 km in north and 3 km in east to get to your school. Then these will be the "**x**" and "**y**" coordinates of your school. The matrix representation of these will look like:

$$
\
\begin{bmatrix}
5 \\
3
\end{bmatrix}
$$

#### Types of Matrices

There are many types of matrices, but the most important ones are:

1. **Row Matrix**: A matrix with only one row. For example, $$ [1, 2, 3] $$
2. **Column Matrix**: A matrix with only one column. For example, $$ \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix} $$
3. **Square Matrix**: A matrix with the same number of rows and columns. For example, $$ \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} $$
4. **Diagonal Matrix**: A matrix with all elements zero except the diagonal elements. For example, $$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 2 & 0 \\ 0 & 0 & 3 \end{bmatrix} $$
5. **Identity Matrix**: A matrix with all diagonal elements as 1 and all other elements as 0. For example, $$ \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix} $$
6. **Scalar Matrix**: A matrix with all diagonal elements as the same number and all other elements as 0. For example, $$ \begin{bmatrix} 2 & 0 & 0 \\ 0 & 2 & 0 \\ 0 & 0 & 2 \end{bmatrix} $$
7. **Singular Matrix**: A matrix whose determinant is 0. For example, $$ \begin{bmatrix} 1 & 2 \\ 2 & 4 \end{bmatrix} $$
8. **Non-Singular Matrix**: A matrix whose determinant is not 0. For example, $$ \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} $$
9. **Transpose of a Matrix**: A matrix formed by converting columns of a matrix into rows and rows into columns. For example, $$ \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}^T = \begin{bmatrix} 1 & 3 \\ 2 & 4 \end{bmatrix} $$
10. **Symmetric Matrix**: A matrix which is equal to its transpose. For example, $$ \begin{bmatrix} 1 & 2 \\ 2 & 3 \end{bmatrix} $$
11. **Skew-Symmetric Matrix**: A matrix which is equal to the negative of its transpose. For example, $$ \begin{bmatrix} 0 & 2 \\ -2 & 0 \end{bmatrix} $$
12. **Orthogonal Matrix**: A matrix whose transpose is equal to its inverse. For example, $$ \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix} $$

### Matrix Addition

Matrix addition is the addition of two matrices. The sum of two matrices is obtained by adding the corresponding elements of the two matrices. The sum of two matrices A and B is a matrix C. Each element of matrix C is the sum of the corresponding elements of matrix A and matrix B.

For example, let's take two matrices A and B:

$$ A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} $$
$$ B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix} $$

The sum of these two matrices will be:

$$ C = \begin{bmatrix} 1+5 & 2+6 \\ 3+7 & 4+8 \end{bmatrix} = \begin{bmatrix} 6 & 8 \\ 10 & 12 \end{bmatrix} $$


### Matrix Subtraction

Matrix subtraction is the subtraction of two matrices. The difference of two matrices A and B is a matrix C. Each element of matrix C is the difference of the corresponding elements of matrix A and matrix B.

For example, let's take two matrices A and B:

$$ A = \begin{bmatrix} 10 & 12 \\15 & 20 \end{bmatrix} $$
$$ B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix} $$

The difference of these two matrices will be:

$$ C = \begin{bmatrix} 10-5 & 12-6 \\ 15-7 & 20-8 \end{bmatrix} = \begin{bmatrix} 5 & 6 \\ 8 & 12 \end{bmatrix} $$


### Matrix Multiplication

Matrix multiplication is the multiplication of two matrices. The product of two matrices A and B is a matrix C. Each element of matrix C is the sum of the product of the corresponding elements of the ith row of matrix A and the jth column of matrix B.

> **Note**:
>
> 1. The number of columns in the first matrix must be equal to the number of rows in the second matrix.
> 2. All Matrices are represented in (Row x Column) format. For example, a 2x3 matrix has 2 rows and 3 columns.
> 3. The resultant matrix will have the same number of rows as the first matrix and the same number of columns as the second matrix.

For example, let's take two matrices A and B with dimensions 2x3 and 3x2 respectively, the important point to note here is that the number of columns in A is equal to the number of rows in B.

$$ A = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix} $$
$$ B = \begin{bmatrix} 7 & 8 \\ 9 & 10 \\ 11 & 12 \end{bmatrix} $$

The product of these two matrices will be:

$$ C = \begin{bmatrix} 1*7+2*9+3*11 & 1*8+2*10+3*12 \\ 4*7+5*9+6*11 & 4*8+5*10+6*12 \end{bmatrix} = \begin{bmatrix} 58 & 64 \\ 139 & 154 \end{bmatrix} $$

You can see the multiplication visually happening on this [website](https://matrixmultiplication.xyz/).


### Determinant of a Matrix

The determinant of a matrix is a scalar value that can be computed from the elements of a square matrix and encodes certain properties of the matrix. The determinant of a matrix A is denoted as |A| or det(A).

To calculate the determinant of a 2x2 matrix, you need to subtract the product of the diagonals. For example, let's take a matrix A:

$$ A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} $$

The determinant of this matrix will be:

$$ |A| = 1*4 - 2*3 = 4 - 6 = -2 $$

To calculate the determinant of a 3x3 matrix, you need to subtract the product of the diagonals of the matrix. For example, let's take a matrix A:

$$ A = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix} $$

The determinant of this matrix will be:

$$
|A| = 1*(5*9 - 6*8) - 2*(4*9 - 6*7) +  3*(4*8 - 5*7) \\
= 1*(45 - 48) - 2*(36 - 42) + 3*(32 - 35) \\
= 1*(-3) - 2*(-6) + 3*(-3) \\
= -3 + 12 - 9 = 0
$$

Fun Fact: Since the determinant of this matrix is 0, it is singular.

Let's see how we can perform these operations in Python.

### Inverse of a Matrix

The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. The inverse of a matrix \( A \) is denoted as \( A^{-1} \).

To calculate the inverse of a matrix, we first need to understand the concept of the _adjoint_ of a matrix:

- The adjoint of a matrix is the transpose of the cofactor matrix of the original matrix.
- The cofactor matrix is formed by calculating the determinant of the submatrices obtained by removing one row and one column for each element.

For example, let's calculate the inverse of a 2x2 matrix \( A \):

$$A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} $$

1. **Determinant**:  
   The determinant of \( A \) is calculated as:

   $$
   |A| = (1 \cdot 4) - (2 \cdot 3) = 4 - 6 = -2
   $$

2. **Adjoint**:  
   The adjoint of \( A \) is the transpose of the cofactor matrix:

   $$
   \text{Adj}(A) = \begin{bmatrix} 4 & -2 \\ -3 & 1 \end{bmatrix}
   $$

3. **Inverse**:  
   The inverse of \( A \) is given by:
   $$
   A^{-1} = \frac{1}{|A|} \cdot \text{Adj}(A)
   $$
   Substituting the values:
   $$
   A^{-1} = \frac{1}{-2} \cdot \begin{bmatrix} 4 & -2 \\ -3 & 1 \end{bmatrix} = \begin{bmatrix} -2 & 1 \\ 1.5 & -0.5 \end{bmatrix}
   $$

For larger matrices, the process involves similar steps but requires more computation.


```python
# Finding the inverse of a matrix
matrix = np.array([[1, 2], [3, 4]])
inverse_matrix = np.linalg.inv(matrix)
print("Original Matrix:")
print(matrix)
print("Inverse Matrix:")
print(inverse_matrix)
```

## Statistics

Let's move on to our next topic, **Statistics**. Statistics is the study of data collection, analysis, interpretation, presentation, and organization. It is a branch of mathematics that deals with the collection, analysis, interpretation, presentation, and organization of data.

Since we are going to learn about neural networks, we will be using statistics to analyze the data and make predictions. We will be using the following statistical concepts:

1. **Mean**: The average of a set of numbers. It is calculated by summing all the numbers and dividing by the count of numbers.
2. **Median**: The middle value of a set of numbers when they are arranged in order. If the count of numbers is even, the median is the average of the two middle values.
3. **Mode**: The value that appears most frequently in a set of numbers. A set of numbers can have one mode, more than one mode, or no mode at all.
4. **Standard Deviation**: A measure of the amount of variation or dispersion in a set of values. It is calculated as the square root of the variance.
5. **Variance**: A measure of how far a set of numbers is spread out from their average value. It is calculated as the average of the squared differences from the mean.
6. **Correlation**: A measure of the relationship between two variables. It can be positive, negative, or zero.
7. **Covariance**: A measure of how much two random variables vary together. It can be positive, negative, or zero.

The list above is not exhaustive, but it covers the most important concepts that we will be using in this course. Let's explore these concepts one by one.


### Mean

One of the most important concepts in statistics is the **mean**. The mean is the average of a set of numbers. It is calculated by summing all the numbers and dividing by the count of numbers.
The mean is a measure of central tendency, which means it gives us an idea of where the center of the data lies. In everyday language, we call it the average.
The formula for calculating the mean is:

$$
\mu = \frac{\sum_{i=1}^{n} x_i}{n}
$$


```python
l = [1, 2, 5, 3, 8, 5, 10, 34]

mean = np.mean(l)
print("Mean:", mean)
```

### Mode

Another measure of central tendency is the **mode**. The mode is the value that appears most frequently in a set of numbers. A set of numbers can have one mode, more than one mode, or no mode at all.
For example, let's take a set of numbers: {1, 2, 2, 3, 4}. The mode of this set is 2 because it appears most frequently. If we take another set of numbers: {1, 2, 3, 4}, this set has no mode because all the numbers appear only once.
If we take another set of numbers: {1, 2, 2, 3, 3, 4}, this set has two modes: 2 and 3 because both appear most frequently.


```python
l = np.array([1, 1, 3, 3, 3 ,3, 5, 5, 5, 5, 5])

mode = l[np.argmax(l)]

print("Mode:", mode)
```

**Explanation**

First we declare a list of numbers. Then, we use the `argmax` function to find the index of the maximum value in the list. Finally, we use that index to get the mode from the list.


### Median

The **median** is the middle value of a set of numbers when they are arranged in order. If the count of numbers is even, the median is the average of the two middle values.
The important point to note is that the median is the central value only after the numbers are sorted. So, we need to sort the numbers first and then find the median.
The formula for calculating the median is:

$$
\text{Median} = \begin{cases} \frac{x_{n/2} + x_{(n/2)+1}}{2} & \text{if n is even} \\ x_{(n+1)/2} & \text{if n is odd} \end{cases}
$$


```python
median = np.median(l)
print("Median:", median)
```

### Standard Deviation

The **standard deviation** is a measure of the amount of variation or dispersion in a set of values.
It shows how much the values deviate from the mean. A low standard deviation means that the values are close to the mean, while a high standard deviation means that the values are spread out over a wider range. In machine learning, if we find standard deviation of some data to be very high, we can say that the data is not good for training the model. So, first we clean and normalize the data and then move further.
The formula for calculating the standard deviation is:

$$
\sigma = \sqrt{\frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}}
$$

$$
\text{where} \quad \mu \quad \text{is the mean of the data.}
$$


```python
a = np.array([23, 45, 33, 67, 88, 44, 11, 32])
b = np.array([10, 11, 12, 13, 11, 9, 10])

dev1 = np.std(a)
dev2 = np.std(b)

print("Deviation 1:", dev1)
print("Deviation 2:", dev2)
```

As clearly visible from above, if the data is too much spread out, the standard deviation will be high. If the data is close to the mean, the standard deviation will be low. You can go ahead and verify this yourself by changing the values of the list and checking the standard deviation.


### Variance

The standard deviation, since it is basically a sum of all +ve and -ve differences from the mean and the value, can be pretty small, therefore, in large datasets it can mean nothing. So, we need to calculate the variance of the data. The variance is a measure of how far a set of numbers is spread out from their average value. It is calculated as the average of the squared differences from the mean.
The formula for variance is basically the same as the standard deviation, but without the square root:

$$
\sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \mu)^2}{n}
$$

$$
\text{where} \quad \mu \quad \text{is the mean of the data.}
$$


```python
var1 = np.var(a)
var2 = np.var(b)
print("Variance 1:", var1)
print("Variance 2:", var2)
```

**Since our standard deviation was already too low, our variance is also low for array `b`**


### Correlation

The **correlation** is a measure of the relationship between two variables. It can be positive, negative, or zero. A positive correlation means that as one variable increases, the other variable also increases. A negative correlation means that as one variable increases, the other variable decreases. A zero correlation means that there is no relationship between the two variables.
The formula for calculating the correlation is:

$$
\text{Correlation} = \frac{\sum_{i=1}^{n} (x_i - \mu_x)(y_i - \mu_y)}{\sqrt{\sum_{i=1}^{n} (x_i - \mu_x)^2}\sqrt{\sum_{i=1}^{n} (y_i - \mu_y)^2}}
$$

$$
\text{where} \quad \mu_x \quad \text{is the mean of x and} \quad \mu_y \quad \text{is the mean of y.}
$$


```python
# corr = np.corrcoef(a, b)
# Will return error since a and b are not of same length
# To find correlation, we need to make sure both arrays are of same length
a = np.array([23, 45, 33, 67, 88, 44, 11, 32])
b = np.array([10, 11, 12, 13, 11, 9, 10, 12])
corr = np.corrcoef(a, b)
print("Correlation Coefficient:", corr)
```

The `corrcoef` function returns a 2D array, where the first row and column represent the correlation of `x` with itself, and the second row and column represent the correlation of `y` with itself. The value at the intersection of the two rows and columns represents the correlation between `x` and `y`. The value is between -1 and 1. A value of 1 means that there is a perfect positive correlation, a value of -1 means that there is a perfect negative correlation, and a value of 0 means that there is no correlation.


### Covariance

The **covariance** is a measure of how much two random variables vary together. It can be positive, negative, or zero. A positive covariance means that as one variable increases, the other variable also increases. A negative covariance means that as one variable increases, the other variable decreases. A zero covariance means that there is no relationship between the two variables.
The formula for covariance is:

$$
\text{Covariance} = \frac{\sum_{i=1}^{n} (x_i - \mu_x)(y_i - \mu_y)}{n}
$$

$$
\text{where} \quad \mu_x \quad \text{is the mean of x and} \quad \mu_y \quad \text{is the mean of y.}
$$


```python
covariance = np.cov(a, b)
print("Covariance Matrix:")
print(covariance)
```

The `cov` function returns a 2D array, where the first row and column represent the covariance of `x` with itself, and the second row and column represent the covariance of `y` with itself. The value at the intersection of the two rows and columns represents the covariance between `x` and `y`. The value is between -1 and 1. A value of 1 means that there is a perfect positive covariance, a value of -1 means that there is a perfect negative covariance, and a value of 0 means that there is no covariance.

## Probability

Probability is the measure of the likelihood that an event will occur. It is a branch of mathematics that deals with the analysis of random events. Probability is used in machine learning to make predictions based on the data. It is also used to evaluate the performance of machine learning models.

The probability of an event is a number between 0 and 1. A probability of 0 means that the event will not occur, while a probability of 1 means that the event will occur. A probability of 0.5 means that the event has an equal chance of occurring or not occurring.

There are different types of probability, but the most important ones are:

1. **Classical Probability**: The probability of an event is calculated based on the assumption that all outcomes are equally likely.
2. **Empirical Probability**: The probability of an event is calculated based on the observed data.
3. **Subjective Probability**: The probability of an event is calculated based on personal judgment or opinion.
4. **Conditional Probability**: The probability of an event given that another event has occurred.
5. **Joint Probability**: The probability of two events occurring together.

We don't need to dive deeper in these concepts, as we will be using them in our machine learning models. But it is important to understand the basic concepts of probability.


First, let's define the probability of an event. The probability of an event is defined as the number of favorable outcomes divided by the total number of outcomes.
The formula for calculating the probability of an event is:

$$
P(A) = \frac{n(A)}{n(S)}
$$

where:

- P(A) is the probability of event A
- n(A) is the number of favorable outcomes for event A
- n(S) is the total number of outcomes in the sample space S
- S is the sample space, which is the set of all possible outcomes
- A is the event whose probability we want to calculate
- n is the number of outcomes


Keep in mind that probability is always between 0 and 1. A probability of 0 means that the event will not occur, while a probability of 1 means that the event will occur. A probability of 0.5 means that the event has an equal chance of occurring or not occurring.
